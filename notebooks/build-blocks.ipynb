```python
with open("../data/inputs.txt") as f:
    mydata = f.read()
print(len(mydata))
print(mydata[:200])
unique_chars = sorted(set(mydata))
print(unique_chars)
vocab_size = len(unique_chars)
```

```python
stoi = {ch:i for i, ch in enumerate(unique_chars)}
itos = {i:ch for i, ch in enumerate(unique_chars)}
```

```python
def encode(s):
    return [stoi[c] for c in s]

def decode(s):
    return [itos[c] for c in s]
```
