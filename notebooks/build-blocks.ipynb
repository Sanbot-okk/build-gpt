```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import random

random.seed(42)
```

```python
with open("../data/inputs.txt") as f:
    mydata = f.read()
# print(len(mydata))
# print(mydata[:200])
unique_chars = sorted(set(mydata))
# print(unique_chars)
vocab_size = len(unique_chars)
```

```python
stoi = {ch:i for i, ch in enumerate(unique_chars)}
itos = {i:ch for i, ch in enumerate(unique_chars)}
```

```python
def encode(s):
    return [stoi[c] for c in s]

def decode(s):
    return [itos[c] for c in s]
```

```python
data = torch.tensor(encode(mydata), dtype=torch.long)
```

```python
train_data = data[:int(0.9*len(data))]
val_data = data[:int(0.1*len(data))]
# decode(train_data[:20].tolist())
```

```python
block_size = 8
batch_size = 4
split="val"
```

```python
# Batch size: B, Block size: T
def get_batch(split):
    data = train_data if split == "train" else val_data
    ix = torch.randint(0, len(data)-block_size, (batch_size,))
    x = torch.stack([data[idx:idx+block_size] for idx in ix]) #(B, T)
    y = torch.stack([data[idx+1:idx+1+block_size] for idx in ix]) # (B, T)
    print(x.shape, y.shape)
    return x,y
```
